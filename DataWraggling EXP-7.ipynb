```python
#27
students = pd.read_csv("students_master.csv")
orig     = pd.read_csv("marks_original.csv")
upd      = pd.read_csv("marks_updated.csv")

orig_i = orig.set_index("Student_ID")
upd_i  = upd.set_index("Student_ID")
cf27 = orig_i.combine_first(upd_i).reset_index()
final27 = pd.merge(students, cf27, on="Student_ID", how="left")
display(final27)
check = pd.DataFrame({
    "Original": orig_i["Marks"],
    "Updated":  upd_i["Marks"],
    "Final":    cf27.set_index("Student_ID")["Marks"],
}).loc[[101, 102]]
display(check)
```

```python
#28
orig = pd.read_csv("marks_original.csv")
upd  = pd.read_csv("marks_updated.csv")

orig_i = orig.set_index("Student_ID")
upd_i  = upd.set_index("Student_ID")
res_a = orig_i.combine_first(upd_i)
res_b = orig_i.combine(upd_i, np.fmax)
side = pd.DataFrame({
    "combine_first": res_a["Marks"],
    "combine_fmax":  res_b["Marks"],
}).sort_index()
side["Differ"] = ~((side["combine_first"] == side["combine_fmax"]) |
                   (side["combine_first"].isna() & side["combine_fmax"].isna()))
display(side)
print("IDs where results differ:", side[side["Differ"]].index.tolist())
```

```python
#29
rt = pd.read_csv("repeat_tests.csv")
pt = rt.pivot_table(index="Student_ID", columns="Subject",
                    values="Marks", aggfunc=["mean", "max"])
print("Original pivot table:")
display(pt)
st = pt.stack()
print("After stack():")
display(st)
print("Row index levels:", st.index.names)
restored = st.unstack().reindex(columns=pt.columns)
print("After unstack():")
display(restored)
print("Restored equals original:", restored.equals(pt))
```
