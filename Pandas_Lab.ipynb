```python
#29
sales = pd.Series([450, 750, 300, 920], index=['Laptop', 'Mouse', 'Keyboard', 'Monitor'])
print(sales)
print(sales.values)
print(sales.index)
print(sales['Mouse'])
print(sales[sales > 500])

sales_dict = sales.to_dict()
print(sales_dict)
```
